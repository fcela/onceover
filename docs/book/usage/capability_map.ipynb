# What onceover can find: the capability map

This page enumerates every kind of statement onceover can make
about a table, grouped by the question the statement answers. It
is the page to read first: the [capabilities guide](capabilities.ipynb)
is about choosing detectors, and the [findings reference](findings.ipynb)
documents each kind's evidence and facets in full. This page
answers the prior question — what is there to find — in one
reading.

The map is generated from the registered taxonomy, the same
export behind `onceover.taxonomy()`, and a freshness test fails
when the page and the taxonomy disagree, so the enumeration
cannot drift from the library. Each registered type appears
exactly once, under the question it answers. A type can have
several forms; the forms and their definitions are listed in the
findings reference. One example statement is given per type, in
the wording the library itself uses.

The Runs column states how the pass that produces the kind is
engaged. *By default*: the pass is in the default detector set.
*On request*: the pass runs only when its detector is named in
`detectors=[...]`. *With declared hints*: the pass is in the
default set but engages only on declared hints. *With a
designated target*: the pass engages when a target is designated
with `target=`. A run records every pass it skipped, and why, in
its coverage.


```python
import onceover

taxonomy = onceover.taxonomy()
types = {pattern["type"] for pattern in taxonomy["patterns"]}
detectors = onceover.available_detectors()
default = sum(1 for d in detectors if d["membership"] == "default_set")
print(f"{len(types)} registered finding types")
print(f"{default} of the {len(detectors)} registered detectors run by default")
```

## Asking for the kinds that do not run by default

Kinds marked *On request* come from detectors outside the
default set; name the detector to run it, for example
`onceover.profile(df, detectors=["row_clusters"])`. Kinds
marked *With a designated target* engage when a target is
designated, for example `onceover.profile(df, target="is_canceled")`.
Kinds marked *With declared hints* engage on the hint that
names their subject: a `contrast` hint for a group contrast,
`dimensions` hints for panel completeness.

Two qualifications apply across the map. The along-axis kinds —
level shifts, trend changes, partitions, shapes — scan the
frame's time axis, or an axis declared with the `ordered_axis`
hint, in a default run; along other numeric axes those detectors
run when named. And a kind's absence from a report is not a
negative result unless the pass's coverage entry says the pass
ran: read [coverage](coverage.ipynb) before concluding that a
table lacks a structure.
