# Further directions

The [capability guide](../usage/capabilities.ipynb) describes implemented
methods. The ideas below extend their scope; they are not scheduled
features or configuration options.

## Broader discovery

**General formulas.** The arithmetic scans already cover affine formulas,
sums, differences, products, ratios, and small linear combinations.
Discovering other nonlinear expressions or searching larger predictor
sets would require additional candidate generation and verification.

**Model-relative anomalies.** Current surprise checks compare a numeric
value with its column or a categorical subset, and categorical values with
nearly universal context patterns. A separate residual-based detector
could identify values that are ordinary marginally but unexpected given
several other measurements. Its validation would need to account for
searching and selecting contexts.

**Larger rule sets and contrast groups.** Nominal association mining
searches pairs, while PRISM and the categorical surprise scan already
build multi-condition contexts. Larger itemsets, self-sufficiency filters,
and explicitly optimized subgroups could answer broader segment-comparison
questions with additional search and selection controls.

**Cross-table structure.** Inclusion dependencies could identify value
containment and lookup relationships between tables. Comparing distributions
between separate datasets is also different from the current structural
`report.diff()` and within-table model-routing drift.

## Time and meaning

**Richer calendars.** Inferred weekday references and declared business-day
cadences do not model regional holidays. Holiday-aware calendars and
subdaily completeness rules would make additional expectations checkable.
Timestamp-ordered breaks and transient excursions already operate within
a day; the missing extension is intraday calendar/gap semantics.

**More change models.** The breaks pass already supports multiple level
breaks and separate trend/variance changes. The reference-model residual
probe tests one breakpoint. Jointly fitting several residual changes,
ramps, or more general segmented models would extend that narrower probe.

**Semantic formats.** Current character signatures and numeric recording
checks expose format differences without interpreting emails, telephone
numbers, currencies, or domain codes. Such semantic recognizers would need
separate validation and maintenance.

## Interfaces and assessment

`target=` currently adds numeric target-shape diagnostics. A response-focused
mode across all discovery passes would be a broader interface change.
Additional predicate transforms, such as month and weekday, would also
extend the current `day_index` representation.

Independent final confirmation, repeated null experiments, sensitivity to
search limits, and task-specific anomaly precision/recall would strengthen
assessment beyond individual fixtures. The experimental views detector
selects one greedy partition; uncertainty over partitions remains a
separate modeling direction.

New passes should record the rows and work limits used, preserve the
measurements behind each claim, and include both planted examples and
negative controls. These requirements make comparisons with existing
methods possible without assuming that a larger report is a better one.
