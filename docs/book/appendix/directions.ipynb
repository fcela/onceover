# Directions

What the profiler could plausibly do next, with the current boundary
drawn honestly in each case. Nothing here is scheduled; everything here
is absent from the current package.

## Discovery

- **Multiple breakpoints.** The residual-shift pass tests one break on
  the first temporal column. Segmented regression with several breaks
  (Muggeo-style iteration, or PELT-style dynamic programming) would
  catch ramp-then-level patterns that one break averages away.
- **Wider itemsets and self-sufficiency.** Association mining is
  bounded to pairs of nominal values. Three-item conditions,
  self-sufficiency filtering (Webb 2010), and significance-corrected
  discovery would widen the rule tier without touching the statement
  schema.
- **Subgroup discovery and contrast sets.** Beam search with WRAcc over
  discretized conditions — the CN2-SD lineage — would generalize the
  "what's different about these rows" question beyond pairs, and
  contrast sets against natural segments (before/after a detected
  change, missing vs. present) would give the changes category a
  follow-up pass.
- **Conditional anomalies.** The GritBot-style blanket pass — a value
  shockingly rare *within* a context — is the most-requested tier-three
  capability and the most delicate to keep honest.
- **Multivariate identities.** The identity probe is pairwise affine
  today; `y ≈ a·x₁ + b·x₂` and product/ratio forms would cover the
  total = quantity × price family the screen currently flags as merely
  predictable.
- **Inclusion dependencies.** Value containment between columns
  (lookup/enumeration relationships) extends the lookup story from
  within-table to between-table, and multi-table profiling generally
  (drift/diff between two profiles) is the natural next boundary.

## Semantics and time

- **Holiday calendars.** Gap semantics are weekday-aware; real calendars
  (regional holidays) would stop flagging public holidays as ETL gaps.
  The current finding says `holidays not modeled` for exactly this
  reason.
- **Intraday structure.** Datetime columns are analyzed at day
  granularity; sessions, time-of-day gaps, and sub-daily cadences are
  open.
- **Semantic patterns.** Roles are behavioral today (identifier-like,
  categorical); pattern libraries for emails, phones, and codes would
  pull in maintenance the package has so far avoided.

## Interfaces

- **Supervised mode.** A `targets=` argument — profile *this* column
  against the rest, with the same passes and the same honesty — is the
  smallest interface change with the largest use case.
- **More predicate transforms.** The rule schema supports the
  `day_index` transform; month, weekday, and window transforms would
  let rules speak calendar.
- **MCP server.** An agent-facing tool wrapper around `profile` so any
  MCP client can call the profiler directly, with the report's
  token-tiered views doing the truncation.
- **Engine seam.** The predictability screen is a purpose-built
  histogram GBM. The statement schema does not care what fits the
  model — a faster engine (or an ensemble) can replace it behind the
  same `dependencies/predictive` statements.

## Discipline

Everything above arrives under the same rules the current passes obey:
bounded compute disclosed in coverage, statements separated from
methods, validation labeled for what it is, and a negative-control
test before any new claim ships.
