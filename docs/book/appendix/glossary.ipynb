# Glossary

| Term | Meaning |
|---|---|
| **AFD** | Approximate functional dependency: a mapping with some exceptions. Reported with a violation count and the g1 error. |
| **basis** | `measured` identifies a count or direct comparison; `estimated` identifies a sample or model estimate; `inferred` identifies an interpretation such as a column role. |
| **budget** | A preset (`quick`, `standard`, or `deep`) for sampling and search limits. Exact column summaries use all rows under every preset. |
| **category** | A label such as `uniqueness`, `dependencies`, or `gaps`; the optional `type` field specifies a subtype. |
| **conditional rate** | Among rows satisfying a rule's conditions, the proportion that also satisfy its outcome. |
| **coverage** | Status, sample size, and limits for each profiling pass. |
| **detector** | A component that performs a check. For example, `dates` produces gap and interval findings. |
| **diagnostic partition** | The fixed one-fifth row subset used to recount conditional rules. Discovery may already have used these rows. |
| **effective cardinality** | The number of equally frequent values with the same Shannon entropy as the observed distribution; calculated as exp(entropy). |
| **evidence** | A finding's method, evaluated rows, validation, and metrics. A finding can include results from several methods. |
| **finding** | An entry in `report.findings`, containing a text summary, the columns involved, and supporting measurements. |
| **form** | A dependency's representation, such as `value_mapping`, `affine`, or `piecewise_linear`. |
| **g1** | Fraction of evaluated rows that would need to change for a functional dependency to hold exactly. |
| **groups** | Findings connected through shared columns. Check individual mappings before deciding that columns are redundant. |
| **holdout** | Rows excluded from fitting and used to evaluate a model. The evidence records their count. |
| **interestingness** | Score used to sort findings. A larger score means higher reporting priority, not a greater probability of correctness. |
| **lift** | A rule's conditional rate divided by its outcome's baseline rate on the same rows. |
| **ξ (xi)** | Chatterjee's rank-based dependence coefficient. The profiler computes both directions because ξ(X,Y) can differ from ξ(Y,X). |
| **scope** | The rows used for a measurement, including input size, evaluated count, and missing-value handling. |
| **statement** | What a finding asserts, identified by its category, type, subject, and specification. Candidates with matching normalized fields are combined. |
| **support** | For a rule, `joint_count` counts rows satisfying both its conditions and its outcome. `joint_support` divides that count by `eligible_rows`. |
| **UCC** | Unique column combination: columns whose joint values uniquely identify the evaluated rows. |
| **validation** | The check used to assess a result, such as direct comparison, cross-validation, or a holdout. It also records whether the check helped select the result. |
