# Glossary

| Term | Meaning |
|---|---|
| **AFD** | Approximate functional dependency: a mapping that holds on most rows; reported with its violation count and the g1 error (violations ÷ rows used). |
| **basis** | The kind of support behind a statement: `measured` (counts and comparisons), `estimated` (fitted, validated on held-out data), `inferred` (an interpretation suggested by measurements). |
| **budget** | The compute tier (`quick`, `standard`, `deep`) that configures every pass at once; the exact work is unchanged across tiers. |
| **category** | The public taxonomy of statements — what was found, independent of which detector found it. |
| **conditional rate** | Among rows matching a rule's conditions, the share that also match its outcome; the evaluator's common unit for rule quality. |
| **coverage** | The per-pass record of what ran, at what scale, and what was skipped or capped — the report's disclosure block. |
| **detector** | A registered discovery component (PRISM, the association miner, the FD pass…) that contributes evidence-bearing statements. Detector names never appear as finding categories. |
| **diagnostic partition** | The deterministic one-fifth row sample on which every conditional rule is re-evaluated; discovery may already have seen those rows, so it validates selection, not truth. |
| **effective cardinality** | exp(Shannon entropy) of a value distribution — "this column behaves like it has N distinct values". |
| **evidence** | One record per contributing method: its algorithm, basis, scope, validation, and raw metrics. |
| **finding** | One statement about the data with its supporting evidence — the ledger's unit. |
| **form** | The mathematical shape of a dependency: `value_mapping`, `affine`, or `piecewise_linear`. |
| **g1** | The approximate-dependency error: the fraction of rows that would need to change for the mapping to hold exactly. |
| **groups** | The connected view over selected findings: which findings share columns. Membership asserts no redundancy. |
| **holdout** | Rows withheld from a fit and used to score it; the sizes travel with the finding. |
| **interestingness** | The deterministic score that ranks the ledger — effect strength weighted by evidence quality, simplicity, and stated form. Not a probability. |
| **lift** | A rule's conditional rate divided by the outcome's baseline rate on the same rows. |
| **ξ (xi)** | Chatterjee's rank-based dependence coefficient; asymmetric, detects arbitrary functional shapes, computed in both directions. |
| **scope** | Which rows a measurement covered: `full`, `sample`, with population and evaluated counts. |
| **statement** | The canonical claim a finding makes — category, type, subject, specification — and the unit of consolidation. |
| **support** | How many rows a rule or relationship actually covers; always paired with the population it was drawn from. |
| **UCC** | Unique column combination: a set of columns whose joint values identify rows. Single-column UCCs are keys; near-unique columns with repeats are dirty keys. |
| **validation** | How a statement was checked: `direct_comparison`, `cross_validation`, `diagnostic_holdout`, or `diagnostic_partition` — plus whether the check participated in selection. |
